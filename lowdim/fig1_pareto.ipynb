# Figure 1 — two-moons, unconstrained: reward vs. control energy

All plotting logic lives in `make_fig1.py` so the figure can also be made headless (`python make_fig1.py`). This notebook runs it and shows the three panels.

* `DATA = "paper_data"` uses the sweep CSVs behind the paper's figure (shipped in this repo).
* `DATA = "results_fig1"` uses a fresh run of `reproduce_fig1.sh`.

```python
import subprocess, sys
from IPython.display import Image, display

DATA = "paper_data"     # or "results_fig1" after running reproduce_fig1.sh
OUT  = "figures"

res = subprocess.run([sys.executable, "make_fig1.py", "--data", DATA, "--out", OUT],
                     capture_output=True, text=True)
print(res.stdout, res.stderr)
res.check_returncode()
```

```python
for name in ("lowdim_lowlambda.png", "lowdim_highlambda.png", "rewards_lowdim.png"):
    print(name)
    display(Image(filename=f"{OUT}/{name}", width=480))
```
